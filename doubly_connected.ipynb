```python
# Calculate generalized Schwarz integral formulas
import importlib
import numpy as np
import matplotlib.pyplot as plt
import maps_core
import context
import alpha_prime
# Reload after editing .py files (one module at a time; dependency order)
importlib.reload(maps_core)
importlib.reload(context)
importlib.reload(alpha_prime)

from context import init_context
from alpha_prime import omega_alpha, X_alpha, dvalpha_dz, dvalpham_dz, G
from alpha_prime import dthetam, thetam, thetam_inv
from maps_core import real, imag, pi, log, sin, cos, exp, prod, conj
from maps_core import is_inside_domain, diff
# selected fixed points (maps_core.build_selected_fps)
from maps_core import fixed_points_of_map_vec

def Phatalpha(z,rho,alpha):
    l = 30
    p = 1
    for li in range(1,l):
        p = p*(1-z*rho**(2*li))**exp(-2j*li*alpha) * (1-1/z*rho**(2*li))**exp(2j*li*alpha)
    return p

def Palpha(z,rho,alpha):
    p = (1-z)*Phatalpha(z,rho,alpha)
    return p
```
