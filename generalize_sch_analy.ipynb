```python
# Domain / Schottky parameters (owned by this notebook)
#
qv = np.array([0.2, 0.1])
#qv = np.array([0.4, 0.3])
#dv = np.array([0.5 , -0.5])
dv = np.array([0.5-0.2j, -0.5+0.1j])
#dv = np.array([0.5-0.2j, -0.5])
#qv = np.array([0.2,0.5,0.1])

alpha = np.array([[np.pi / 5], [np.pi / 3]])
#alpha = np.array([[np.pi / 2], [np.pi / 4]])
level = 10

init_context(qv, dv, alpha, level_in=level)

def dlogomega_dz(z,a):
    return 1/2*dlogX_dz(z,a)
```

```python
def f(z):
    return z

Nd = 2000
# Nd = 2000
dth = 2*pi/Nd
zp = exp(1j*np.linspace(dth/2,2*pi-dth/2,Nd,endpoint=True))
ea = exp(1j*alpha)
eam = exp(-1j*alpha)

qzp0 = qv[0]*zp + dv[0]
qzp1 = qv[1]*zp + dv[1]
## --- boundary values --- 
r0 = real(f(zp))
r1 = real(ea[0]*f(qzp0))
r2 = real(ea[1]*f(qzp1))
a = 0.1+0.3j

print(is_inside_domain(a, qv, dv))

# explicit calculation
kerC0n1 = 1j*dlogomega_dz(zp,a)*zp
kerC0n2 = -1j*conj(dlogomega_dz(zp,1/conj(a))*zp)
kerC1n1 = 1j*dlogomega_dz(qzp0,a)*zp*qv[0]
kerC1n2 = -1j*conj(dlogomega_dz(qzp0,1/conj(a))*zp*qv[0])
kerC2n1 = 1j*dlogomega_dz(qzp1,a)*zp*qv[1]
kerC2n2 = -1j*conj(dlogomega_dz(qzp1,1/conj(a))*zp*qv[1])


val0n = sum(r0 * (kerC0n1 + kerC0n2))/(2*pi*1j)*dth
val1n = sum(r1 * (eam[0]*kerC1n1 + ea[0]*kerC1n2))/(2*pi*1j)*dth
val2n = sum(r2 * (eam[1]*kerC2n1 + ea[1]*kerC2n2))/(2*pi*1j)*dth
print(f(a))

cj = 0
c0_0 = sum(r0*dvalpha_dz(zp,alpha,cj)*1j*zp)*dth
c0_1 = eam[0]*sum(r1*dvalpha_dz(qzp0,alpha,cj)*1j*qv[0]*zp)*dth
c0_2 = eam[1]*sum(r2*dvalpha_dz(qzp1,alpha,cj)*1j*qv[1]*zp)*dth
c0 = 1/sin(alpha[cj])*(c0_0 - c0_1 - c0_2)

print(val0n-val1n-val2n + 1j*real(c0))
error2 = abs((f(a) - (val0n-val1n-val2n + 1j*real(c0)))/f(a))
print(error2)
```

```python
# single-valued condition
def dmu_dz(z,m):
    return dvalpha_dz(z,alpha,m)/sin(alpha[m]) - dvalpha_dz(z,alpha,0)/sin(alpha[0])

s0 = sum(r0*dmu_dz(zp,1)*zp)*dth
s1 = eam[0]*sum(r1*dmu_dz(qzp0,1)*zp)*dth*qv[0]
s2 = eam[1]*sum(r2*dmu_dz(qzp1,1)*zp)*dth*qv[1]

print(s0 - s1 - s2)





```
